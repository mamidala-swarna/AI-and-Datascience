```python

#for loop ->10000 products
prices_of_products=[1000,1500,2000,2500,3000,3500,5000,10000]
#for a variable in sequence
#statements
print("Prices Before Discount")
for individual_product_price in prices_of_products:
    print(individual_product_price)
print("Prices After Discount")
for individual_product_price in prices_of_products:
    print(individual_product_price-250)       
```

### Nested Loops ###

```python
#Nested Loops
for outer in range(1,6,1):
    print(outer)
for inner in range(1,6,1):
    print(inner)  
```

```python
for outer in range(1,6,1):
    for inner in range(1,6,1):
        print(f"{outer}X{inner}={outer*inner}")
```

```python
#Nested loop Real World use cases
colors=["red","black","blue"]
sizes=["uk-7","uk-8","uk-9","uk-11"]
for color in colors:
  for size in sizes:
       print(color +"-----------"+size)
```

### Branching Structures ###

```python
#Statements(Jump Statements)
#stop when num is 5
#break:helps you to exit the loop
for num in range(1,11,1):
    if num==5:
        break
    print(num)
```

```python
#skip 5th number
#continue:helps you skip the current iteration
for num in range(1,11,1):
    if num==5:
        continue
    print(num)
```
