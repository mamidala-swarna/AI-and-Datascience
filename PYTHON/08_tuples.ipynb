```python
#First Element
data = (10,20,30,40,50)
first_element=data[0]
print("First Element",first_element)

#Last Element
last_element=data[-1]
print("Last Element",last_element)

# unknown_element = data[10] # IndexError: tuple index out of range
# print(unknown_element)
```

```python
# Slicing in Tuples is same as Strings 
data = (10,20,30,40,50)
print(data)
print(data[::]) # 10, 20, 30, 40, 50
print(data[1:3:]) # 20,30
print(data[0:5:2]) # 10,30,50
```

```python
# Access Individual Elements 
data = (10,20,30,40,50)
print(data[0])
print(data[1])
print(data[2])
print(data[3])
print(data[4])
```

```python
# Access Individual Elements  -> 10k elements 
data = (10,20,30,40,50,1000000)
print(data[0])
print(data[1])
print(data[2])
print(data[3])
print(data[4])
# print(data[9999999])
```

```python
#Access Individual elements ->10k elemnts
data=(10,20,30,40,50,1000000)
#print(dir(data)) #__iter__
for num in data:
    print(num)


#Apply Operators ->Requirement:Give Courses In upper case
data=("python","ai","cloud")
print(data)
for course in data:
    print(course.upper())
```

```python
#Apply conditionals ->Requirement:Give only even numbers
data=(10,20,30,40,55)
for num in data:
    if num%2==0:
        print(num)
```

```python
# Tuple Methods / Operations 

# index(): Used To Get Index Position Of Given Element
data = (10,20,30,40,50)
print(data)
data.index(30)
print(data.index(30))
print(data.index(50))

# count():  Count the number of occurrences of a value
data = (10,20,10,30,10,40,10)
print(data)
data.count(10)
print(data.count(10))
```
