```python
#operators
#Arithmetic Operators
num1=10
num2=5
print("Sum of Numbers: ",num1+num2)
print("Difference of Numbers: ",num1-num2)
print("Product of Numbers: ",num1*num2)
print("Division of Numbers",num1/num2)
print("Modulus of Numbers: ",num1%num2)

print("Normal Division: ",3/2)
print("Floor Division: ",3//2) #removed point value #roundof
print("Exponentiation: ",3**2)
```

```python
#Compound Assignment Operators
num=10
num=num+5
print(num)

num=10
num+=5
print(num)
```

```python
count = 0
print(count)
# count++ # SyntaxError: invalid syntax
count += 1
print(count)
count += 1
print(count)
count += 1
print(count)

count = 10
print(count)
count -= 1
print(count)
count -= 1
print(count)
```

```python
#Comparision Operators
num1=3
num2=2
print(num1==num2)
print(num1!=num2)
print(num1>num2)
```

```python
#Logical Operators
num1=1
num2=2
num3=3
num4=4
print(num1>num2 and num3>num4)#F and F->F
print(num1<num2 and num3>num4)#T and F->F
print(num1<num2 and num3<num4)#T and T->T

print(num1>num2 or num3>num4)#F or F->F
print(num1<num2 or num3>num4)#T or F->F
print(num1<num2 or num3<num4)#T or T->T

print(num1>num2) #F
print(not num1>num2 )#T
```

```python
#Membership operators
sentence="python is interpreted language"
find_word="java"
status=find_word in sentence
print(status)

sentence="python is interpreted language"
find_word="python"
status=find_word in sentence
print(status)
```

```python
#List Data Type-[]
employee_ids=["101101","202101","301101","901101"]
find_by_emp_id="701101"
status=find_by_emp_id in employee_ids
print(status)

status=find_by_emp_id not in employee_ids
print(status)
```

```python
#Identity Operators
n1=10
n2=10
n3=5
print(id(n1))
print(id(n2))
print(id(n3))

print(n1 is n3)
print(n1 is n2)
print(n1 is not n3)


n1=[10,20,30]
n1=[10,20,30]
print(id(n1))
print(id(n2))
print(n1 is n2)
```

```python
#Bitwise Operators
n1=5
n2=3
print(n1&n2)#1
print(n1|n2)#7
```

```python
#Type Conversion/Implicit Conversion[Automatic]
n1=10#int
n2=5.5#float
sum=n1+n2
print(sum)
print(type(sum))
```

```python
#Type Casting/Explicit Conversion[Manual]
price=1125.890#float
print(price)
print(type(price))

#round_off_price=data_type(variable_name)
round_off_price=int(price)
print(round_off_price)
print(type(round_off_price))
```
