```python
#Functional Programming
#without functions
#user wants to do calculations with below values
num1=10
num2=5
#math operations
print(num1+num2) #addition
print(num1-num2) #subtraction
print(num1*num2) #multiplication
print(num1/num2) #division
```

```python
#user wants to do calculations with below values
num1=20
num2=5
#math operations
print(num1+num2) #addition
print(num1-num2) #subtraction
print(num1*num2) #multiplication
print(num1/num2) #division
```

```python
#user wants to do calculations with below values
num1=200
num2=50
#math operations
print(num1+num2) #addition
print(num1-num2) #subtraction
print(num1*num2) #multiplication
print(num1/num2) #division
```

```python
#with functions
def math_ops():
    print(num1+num2) #addition
    print(num1-num2) #subtraction
    print(num1*num2) #multiplication
    print(num1/num2) #division
#user wants to do calculations with below values
num1=10
num2=5
math_ops()
#user wants to do calculations with below values
num1=20
num2=5
math_ops()
#user wants to do calculations with below values
num1=200
num2=50
math_ops()
```

```python
#math_ops(1000,500)#math_ops(1000,500)
#with functions and parameters
def math_ops(num1,num2):
    print(num1+num2) #addition
    print(num1-num2) #subtraction
    print(num1*num2) #multiplication
    print(num1/num2) #division
math_ops(10,5)#user wants to do calculations with below values
math_ops(20,5)#user wants to do calculations with below values  
math_ops(200,50)#user wants to do calculations with below values    
```

```python
#positional arguments
def employee_info(emp_name,emp_email,emp_location):
    print(f"Hi {emp_name}, your email is {emp_email} and you are located in {emp_location}")
employee_info("Hyderabad","swarna","swarnamamidala@gmail.com")#order matters leading to errors or unexpected behaviour
employee_info("swarna","swarnamamidala@gmail.com","Hyderabad")#order matters leading to errors or unexpected behaviour
```

```python
#keyword arguments
def employee_info(emp_name,emp_email,emp_location):
    print(f"Hi {emp_name}, your email is {emp_email} and your work location is {emp_location}")
employee_info("Hyderabad","swarna","swarnamamidala@gmail.com")#positional
employee_info(emp_location="Hyderabad",emp_name="swarna",emp_email="swarnamamidala@gmail.com")#keyword
```

```python
# Without Default Arguments
def employee_info(emp_name,emp_email,emp_location,org_name):
    print(f"Hi {emp_name} your email is {emp_email} and working for {org_name} at location {emp_location}")

employee_info(emp_location="Hyderabad",emp_name="ravi",emp_email="ravi2krishna@gmail.com",org_name="TCS") # Keyword 
employee_info(emp_location="Bangalore",emp_name="krishna",emp_email="krishna@gmail.com",org_name="TCS") # Keyword 
employee_info(emp_location="Pune",emp_name="Sam",emp_email="sam@gmail.com",org_name="TCS") # Keyword 
# Similarly we have 20 other employees all working for TCS
```

```python
# With Default Arguments
def employee_info(emp_name,emp_email,emp_location,org_name="TCS"):
    print(f"Hi {emp_name} your email is {emp_email} and working for {org_name} at location {emp_location}")

employee_info(emp_location="Hyderabad",emp_name="ravi",emp_email="ravi2krishna@gmail.com") # Keyword 
employee_info(emp_location="Bangalore",emp_name="krishna",emp_email="krishna@gmail.com") # Keyword 
employee_info(emp_location="Pune",emp_name="Sam",emp_email="sam@gmail.com") # Keyword 
# Similarly we have 20 other employees all working for TCS 
```
