```python
import numpy as np
u=np.array([1,2,3])
v=np.array([4,5,6])
print("u+v",u+v) #element-wise addition
```

```python
print("u-v",u-v)#element wise substraction
```

```python
k=6
print("k*u =",k*u)
```

```python
print("Magnitude of u =",np.linalg.norm(u))
```

```python
#unit vector=>a vector whose magnitude is 1
k=np.linalg.norm(u)
unit_vector_u=u/k
print(np.linalg.norm(unit_vector_u))
```

```python
#Matrix
#A=[[1 2]
#   [3 4]]
#shape of a matrix=>2x2=>no.of rows x no.of columns

A=np.array([[1,2],[3,4]])
B=np.array([[5,6],[7,8]])

print(A)
print(A.shape)
print(B)
print(B.shape)
```

```python
A=np.array([[1,2],[3,4]])
B=np.array([[5,6,9],[7,8,10]])

print(A.shape)
print(B.shape)
```

```python
C=A@B
print(C.shape)
print(C)
```

```python
#Transpose=>rows and columns are flipped
A= np.array([[1,2],[3,4]])
B = np.array([[5,6,9],[7,8,10]])

print(A)
print("Transpose of A:",A.T)

print(B)
print("Transpose of B:",B.T)
```

```python
C=B.T
print(C)
```

```python
print(C@A)
```
