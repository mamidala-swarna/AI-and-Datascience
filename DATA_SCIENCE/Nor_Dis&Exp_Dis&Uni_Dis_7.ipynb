```python
# PDF (Probability Distribution function):


#1) Normal Distribution(Gaussian Distribution):

#-> Normal distribution is a widely used continuous probability distribution in statistics and ML . It is a bell-shaped
#curve symmetric about mean.

#PDF: f(x)= (1/ (sigma*sqrt(2* pie))) * e^(-(x-mu)^2/ 2* sigma^2)
#sigma -> standard deviation (controls speed)
#mu -> mean (centre of the curve)


#properties:
#-> symmetric around mean(m)
#-> total area under the curve will be 1
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mu=0
sigma=1
#standard normal distribution

x=np.linspace(-4,4,1000)
y=norm.pdf(x,mu,sigma)
plt.plot(x,y,label="Normal PDF")

plt.axvline(-1,color='red',linestyle='--',label="x=-1")
plt.axvline(1,color='blue',linestyle='--',label="x=+1")

plt.fill_between(x,y,where=(x>-1)&(x<1),alpha=0.3,color='green')

plt.title("Standard Normal Distribution")
plt.xlabel("x")
plt.ylabel("Density")
plt.legend()
plt.show()
```

```python
# Exponential distribution:

#=>exponential distribution models the time between events in a poisson proces(events occur independently at a constant average rate)

# PDF:f(x;lam)=lam*e^(-lam*x)
# mean=1/lam
```

```python
from scipy.stats import expon
lam=1/2 #mean

x=np.linspace(0,10,1000)
y=expon.pdf(x,scale=1/lam)

plt.plot(x,y,label="Exponential PDF")
plt.title("Exponential distribution (lam=0.5)")
plt.xlabel('x')
plt.ylabel('Density')
plt.legend()
plt.show()
```

```python
# Uniform distribution
# -> uniform distribution assumes all outcomes are uniformly likely in a given range

#pdf f(x)=1/b-a   a<=x<b
#mean ->(a+b/2)
```

```python
from scipy.stats import uniform
a=0
b=10
x=np.linspace(a,b,1000)
y=uniform.pdf(x,a,b-a)

plt.plot(x,y,label="Uniform PDF")

plt.title("Uniform Distribution(0,10)")
plt.xlabel('x')
plt.ylabel('Density')
plt.legend()
plt.show()
```

```python
# 1. The exam scores of students follow a Normal distribution with μ=70,σ=10.
#     - What proportion of students score **between 60 and 80**?
# 2. The lifetime (in hours) of a lightbulb follows an Exponential distribution with λ=1/200.
#     - What is the probability that a bulb lasts **more than 500 hours**?
# 3. A number is chosen randomly between 5 and 15 (Uniform distribution).
#     - What is the probability that the number lies **between 7 and 12**?
```

```python
# 1.Normal Distribution
from scipy.stats import norm
mu=70
sigma=10
prob=norm.cdf(80,mu,sigma)-norm.cdf(60,mu,sigma)
print("Probability of students scoring between 60 and 80:",prob)
```

```python
#2.Exponential Distribution 
from scipy.stats import expon
lam=1/200
prob=expon.sf(500,scale=1/lam)
print("Probability that the bulb lasts more than 500 hours:",prob)
```

```python
# 3.Uniform Distribution
from scipy.stats import uniform
a=5
b=15
prob=uniform.cdf(12,a,b-a)-uniform.cdf(7,a,b-a)
print("Probability that the number lies betwen 7 and 12:",prob)
```
