```python
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score


# sample dataset

data = {
    "text": [
        "claim your free gift card now",
        "your lottery ticket has won a prize",
        "team meeting is scheduled for 10 AM",
        "please submit the report by Friday",
        "limited time discount available today",
        "can we review the project tomorrow",
        "you have been selected for a cash reward",
        "urgent offer click here to receive money",
        "the client meeting starts at 3 PM",
        "don't forget to update the project document",
        "congratulations you won a free vacation",
        "get exclusive deals on your favorite products",
        "let's discuss the quarterly results",
        "please send me the meeting notes",
        "win a brand new smartphone today",
        "special bonus waiting for you claim now",
        "the development team will meet after lunch",
        "your assignment deadline is next Monday",
        "click now to receive your free reward",
        "we need to discuss the project progress"
    ],
    "label": [
        "spam",
        "spam",
        "ham",
        "ham",
        "spam",
        "ham",
        "spam",
        "spam",
        "ham",
        "ham",
        "spam",
        "spam",
        "ham",
        "ham",
        "spam",
        "spam",
        "ham",
        "ham",
        "spam",
        "ham"
    ]
}


df = pd.DataFrame(data)

# input and output
X_text = df['text']
y = df['label']

# tf-idf
vectorizer = TfidfVectorizer()

X= vectorizer.fit_transform(X_text)

print('Feature names:')

print(vectorizer.get_feature_names_out())

print('Tf-idf Matrix')
print(X.toarray())
```

```python
# train-test split

X_train, X_test, y_train, y_test = train_test_split(
    X,y, test_size=0.3,random_state=42
)

# Model
nb = MultinomialNB()
nb.fit(X_train,y_train)

# predictions
y_pred = nb.predict(X_test)

print('Predictions ', y_pred)

print('Actual:', y_test.values)

print("Accuracy:", accuracy_score(y_test,y_pred))
```

```python
import pandas as pd

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import GaussianNB

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score


# sample dataset

data = {
    "text": [
        "claim your free gift card now",
        "your lottery ticket has won a prize",
        "team meeting is scheduled for 10 AM",
        "please submit the report by Friday",
        "limited time discount available today",
        "can we review the project tomorrow",
        "you have been selected for a cash reward",
        "urgent offer click here to receive money",
        "the client meeting starts at 3 PM",
        "don't forget to update the project document",
        "congratulations you won a free vacation",
        "get exclusive deals on your favorite products",
        "let's discuss the quarterly results",
        "please send me the meeting notes",
        "win a brand new smartphone today",
        "special bonus waiting for you claim now",
        "the development team will meet after lunch",
        "your assignment deadline is next Monday",
        "click now to receive your free reward",
        "we need to discuss the project progress"
    ],
    "label": [
        "spam",
        "spam",
        "ham",
        "ham",
        "spam",
        "ham",
        "spam",
        "spam",
        "ham",
        "ham",
        "spam",
        "spam",
        "ham",
        "ham",
        "spam",
        "spam",
        "ham",
        "ham",
        "spam",
        "ham"
    ]
}



df = pd.DataFrame(data)

# input and output
X_text = df['text']
y=df['label']

# count vectorizer
vectorizer=CountVectorizer()
X= vectorizer.fit_transform(X_text)

# convert sparse matrix to dense

X_dense = X.toarray()

print('count matrix:')
print(X_dense)



# train-test split

X_train, X_test, y_train, y_test = train_test_split(
    X_dense,y, test_size=0.3,random_state=42
)

# Model
nb = GaussianNB()
nb.fit(X_train,y_train)

# predictions
y_pred = nb.predict(X_test)

print('Predictions ', y_pred)

print('Actual:', y_test.values)

print("Accuracy:", accuracy_score(y_test,y_pred))
```
