```python
#Probability Distributions:

#1)Binomial Distribution
#->The binomial distribution models of successes in a fixed number of independent trails, each with same probability of success

#conditions for Binomial:
#1) fixed number of trials n
#2) Each trial has two outcomes: success(p) or failure(1-p)
#3) Trails are independent

#PMF(probability mass function):
# P(X=k)= (nck) * p^k * (1-p)^(n-k)

#  10 -> success is seeing number 3
# P(X=7) = 10c7 * (1/6)^7 * (5/6)^3

```

```python
from scipy.stats import binom
n=10
p=0.5
k=5
prob=binom.pmf(k,n,p)
print(prob)
```

```python
import matplotlib.pyplot as plt

X=range(0,n+1)
y=binom.pmf(X,n,p)

plt.bar(X,y)
plt.title("Binomial Distribution(n=10,p=0.5)")
plt.xlabel("number of successes(k)")
plt.ylabel("probability")
plt.show()
```

```python
#Poisson Distribution:
## Defnition: It models the number of events occuring in a
# fixed interval of time or space,when events happen independently and at a constant rate

# P(X=k) = lambda^k * e^(-lambda) / k!

# Example: if a call centre receives 5 calls per minute on average(lambda=5), then what is the probability that exactly 3 calls come in a minute?

# P(X=3) = 5^3 * e^(-3) / 3!
```

```python
from scipy.stats import poisson
lam=5
k=3
prob=poisson.pmf(k,lam)
print(prob)
```

```python
lam=5
x=range(0,15)
y=poisson.pmf(x,lam)
plt.bar(x,y)
plt.title("Poisson Distribution(lambda=5)")
plt.xlabel("Number of events(k)")
plt.ylabel("probability")

plt.show()
```

```python
#Geometric distribution:

#It models the number of trials until the first success in a sequence of independent trials(each with probability p)

# P(X=k) = (1-p)^(k-1) * p
```

```python
# if the probability of hitting an target = 0.2, whats the probability that the first success occurs on the 4rth attempt

# p(X=4) =(1-0.2)^3 * 0.2
```

```python
from scipy.stats import geom
p=0.2
k=4
prob=geom.pmf(k,p)
print(prob)
```

```python
#visualization

x=range(1,15)
y=geom.pmf(x,p)

plt.bar(x,y)
plt.title("Geometric Distribution(p=0.2)")
plt.xlabel("Trial number of first success(k)")
plt.ylabel("Probability")
```

```python
# 1. A factory machine produces **1 defective item per 20 items** on average.
#     - Model this with a **Binomial distribution** (n=50, p=0.05).
#     - Find the probability that exactly 3 items are defective.
# 2. On average, a bookstore sells **2 books per hour**.
#     - Model this with a **Poisson distribution** (λ=2).
#     - Find the probability that **no books are sold** in the next hour.
# 3. A basketball player has a **60% chance of making a free throw**.
#     - Model this with a **Geometric distribution**.
#     - Find the probability that the player makes the first successful shot on the **3rd attempt**.
```

```python
# 1.Binomial Distribution
# Probability of exactly 3 defective items out of 50, with p = 0.05:

from scipy.stats import binom
n=50
p=0.05
k=3
prob=binom.pmf(k,n,p)
print("Probability of exactly 3 defective items:",prob)
```

```python
# 2.Poisson Distribution
# Probability that 0 books are sold when the average is 2 books/hour

from scipy.stats import poisson
lam=2
k=0
prob=poisson.pmf(k,lam)
print("Probability of no books are sold:",prob)
```

```python
#3.Geometric Distribution
#Probability that the first successful free throw occurs on the 3rd attempt,with p=0.6

from scipy.stats import geom
p=0.6
k=3
prob=geom.pmf(k,p)
print("Probability of first success on 3rd attempt:",prob)
```
