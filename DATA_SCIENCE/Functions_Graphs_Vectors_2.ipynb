```python
#Linear Function
#y=mx+c
#m=slope(rate at change)
#c=intercept(the point at which the line meets y-axis)

#plotting linear graph:

import numpy as np
import matplotlib.pyplot as plt

x=np.linspace(-10,10,200)
#print(x)

y=2*x+5

plt.plot(x,y,label="y=2x+5")
plt.title("Linear Function")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.legend()#helpful to place the legend on the axes
plt.show()
```

```python
#Polynomial Function

y_quad=x**2-4*x+3
y_cubic=x**3-6*x**2+11*x-6
x=np.linspace(-10,10,200)

plt.plot(x,y_quad,label="y=x^2-4x+3")
plt.plot(x,y_cubic,label="x^3-6x^2+11x-6")
plt.title("Polynomial Functions")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.show()
```

```python
#Exponential Function

y_growth=2*(1.5**x)
y_decay=100*(0.8**x)

plt.plot(x,y_growth,label="Growth")
plt.plot(x,y_decay,label="Decay")
plt.title("Exponential Function")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.show()
```

```python
#Logarithmic Function

x_pos=np.linspace(0.1,10,200)#x must be greater than 0
y_log=np.log(x_pos)#natural log->log to the base e
y_log10=np.log10(x_pos)#base 10 log

plt.plot(x_pos,y_log,label="ln(x)")
plt.plot(x_pos,y_log10,label="log10(x)")
plt.title("Logarithmic Function")
plt.xlabel("x")
plt.ylabel=("y")
plt.grid()
plt.legend()
plt.show()
```

```python
# Linear -> straight line

# polynomial -> curves and bends

# Exponential -> Rapid growth /decay  -> population

# logarithmic -> slow growth, inverse of exponential
```
