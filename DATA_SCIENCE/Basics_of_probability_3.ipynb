```python
#  Sample Space=> sample space is a set of all possible outcomes of an experiment
#  Example:rolling of a dice

#  S={1,2,3,4,5,6}

#  example: pack of cards
#  S={52 cards}

#  Event=>subset of a sample space

#  Rolling a dice:
#       Event A:"Getting an even number"->A={2,4,6}
#       Event B:"Geting an prime number"->B={2,3,5}

#    Certain event->whole sample space(probability of occurance of that event is 1)
#    Example:probability of occurance of a positive number->6/6=1

#    Impossible event->empty set(probability of occurance of that event is 0)
#     Example:probability of occurance of a number > 6
#     probaility of occurance a negative number

```

```python
#  Axioms
# 1) Non-negativity:
# P(E)>=0
# probability of an event is non negative

# 2) Normalization:
# P(S)=1
# probability of sample space is 1

# 3) Additive(Disjoint Events):
# A={1,3,5}
# B={2,4}
# If A and B are mutually exclusive(cannot happen together)
# P(A U B)=p(A) + P(B)

# P(A) = 3/6
# P(B) = 2/6
# C = A U B = {1,2,3,4,5}
# P(C) =  5/6
```

```python
# 3. Write Python code to **simulate 10,000 rolls of 2 dice** and estimate the probability of the sum being **7**.
#  Compare with theoretical probability.

import random
#no.of rolls
n=10000

#counter for sum=7
count=0

#simulate 10,000 rolls
for i in range(n):
    die1=random.randint(1,6)
    die2=random.randint(1,6)

    total=die1+die2

    if total==7:
        count+=1
#Estimated probability
estimated_probability=count/n

# Theoretical probability
# There are 36 possible outcomes when rolling 2 dice
# Six outcomes have a sum of 7:
# (1,6), (2,5), (3,4), (4,3), (5,2), (6,1)

theoretical_probability=6/36

#Display results
print("Total number of rolls:",n)
print("Number of times sum is 7:",count)
print("Estimated Probability:",estimated_probability)
print("Theoretical Probability:",theoretical_probability)



```
