```python
#A Random Variable(RV) is a function that assigns a numerical value to the outcome of a random experiment
```

```python
#Types of Random Variables

#1)Discrete Random Variables(DRV):

#1,2,3,4,5,6,7,8,9
#Discrete takes countable values(finite or countable infinite)
#Examples:
# Tossing a coin->(X={0=tails,1=heads})
# Rolling a die->(X={1,2,3,4,5,6})
# number of customers arriving in an hour

#2)Continuous Random Variables(CRV):

#1,1.01,1.02,....
#Take values in an interval(uncountable infinite)
#Example:
#  height of a person(continuous range)
#  time taken for completion of a task
#  temperature in a city
```

```python
# Probability Mass Function(PMF):

# for siscrete variables,probabilities are represented using a PMF
# PMF assigns probabilities to each possible values:
# P(X=x)=p(x) where summation(all possible x values ->p(x))=1
# ->The probability of a single outcome is positive
# ->Total probability acrosss all outcomes=1

#example:rolling of a dice
#PMF:
#p(x)=1/6,where x=1,2,3,4,5,6
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6]
p=[1/6]*6
print(x,p)
```

```python
#stem plot
plt.stem(x,p)
plt.title('PMF of a Die Roll')
plt.xlabel("Outcome")
plt.ylabel("probability")
plt.show()
```

```python
# PDF (probability Density Function)

# For continuous random variables, probability are represented using a pdf
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

#normal distribution

x=np.linspace(-4,4,1000)
pdf=norm.pdf(x,0,1) #mean=0 , std=1

plt.plot(x,pdf,label="PDF(Norm(0,1))")
plt.fill_between(x,pdf,alpha=0.2)
plt.title("Probability Density Function")
plt.xlabel("x")
plt.ylabel("Density")
plt.legend()
plt.show()
```
