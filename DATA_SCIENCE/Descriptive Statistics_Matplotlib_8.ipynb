```python
#Descriptive Statistics:
#Central tendency of the data:Mean,Median,Mode
# Mean-average of the data
# Median-1 2 3 4 5 5.5 6
# []
# 0 7 8 9 
# Mode - most frequenctly occuring data
# IQR->Inter Quartile Range->Measures spread of middle 50 percent of the data
# ->Less sensitive outliers 
```

```python
import numpy as np
data=[2,4,6,8,10]
import matplotlib.pyplot as plt
plt.boxplot(data)
plt.title("Boxplot showing median and IQR")
plt.show()
```

```python
#Basic Line plot
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(0,10,1000)
y=np.sin(x)
fig,ax=plt.subplots(figsize=(10,5))
ax.plot(x,y,color='steelblue',linewidth=2)
ax.set_title('sine_wave')
ax.set_xlabel('x')
ax.set_ylabel('sin(x)')

ax.grid(True,alpha=0.1)
plt.show()
```

```python
#multiple lines + legend 
x=np.linspace(0,10,200)
fig,ax=plt.subplots(figsize=(10,5))
ax.plot(x,np.sin(x),label='sin(x)')
ax.plot(x,np.cos(x),label='cos(x)')
ax.set_title('Sine vs Cosine')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.legend()
ax.grid(True,alpha=0.3)
plt.show()
```

```python
#Line Styles
x=np.arange(1,6)
fig,ax=plt.subplots(figsize=(8,4))
#y=x
ax.plot(x,x,'r-',label='solid')
ax.plot(x,x+1,'b--',label='dashed')
ax.plot(x,x+2,'g-.',label='dosh-dot')
ax.plot(x,x+3,'m:',label='dotted')
ax.set_title("Line Styles")
ax.legend()
ax.grid(True,alpha=0.3)
plt.show()
```

```python
#Scatter plot
#Basic scatter plot
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
x=np.random.randn(100)
y=np.random.randn(100)

fid,ax=plt.subplots(figsize=(7,5))
ax.scatter(x,y,alpha=0.9)
ax.set_title('Random Scatter')
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.grid(True,alpha=0.5)
plt.show()
```
