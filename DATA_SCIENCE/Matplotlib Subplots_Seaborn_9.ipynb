```python
#Multiple subplots
import matplotlib.pyplot as plt
import numpy as np

#lets create 2x2 grid
fig,axes=plt.subplots(2,2,figsize=(12,10))

x=np.linspace(0,10,100)#generate values

#0 0
#0 0

#top left subplot(row 0,col 0)

axes[0,0].plot(x,np.sin(x))
axes[0,0].set_title("Sine")

axes[0,1].plot(x,np.cos(x),'r')
axes[0,1].set_title('Cosine')
#axes[0,1].set_ylabel('cos(x)')

plt.tight_layout()# Adjust spacing to prevent overlap
plt.show()
```

```python
#Multiple subplots
import matplotlib.pyplot as plt
import numpy as np

#lets create a 2x2 grid
fig,axes=plt.subplots(2,2,figsize=(12,10))

x=np.linspace(0,10,100)#generate values
#0 0
#0 0
#top left subplots(row 0 ,col 0)
axes[0,0].plot(x,np.sin(x))
axes[0,0].set_title('Sine')

#top right
axes[0,1].plot(x,np.cos(x),'r')
axes[0,1].set_title('Cosine')

#bottom left
axes[1,0].plot(x,np.tan(x))

#bottom right
axes[1,1].plot(x,np.exp(-x/10))

plt.tight_layout() #Adjusting spacing to prevent overlap

plt.show()
```

```python
# Seaborn -> Its a python data visualization library built on top of Matplotlib and its designed for Statistical graphics

# Advantages

# -> Beautiful default styles
# -> built in themes
# -> statistical functions integrated
# -> works seamlessly with Pandas DataFrames
# -> simpler syntax for complex plots
# -> better color palettes
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips=sns.load_dataset('tips')
print(tips)

fig,ax=plt.subplots(figsize=(10,6))
sns.histplot(data=tips,x='total_bill',kde=True,ax=ax)
#kde->True #adds smooth density curve overlay
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips=sns.load_dataset('tips')
fig,ax=plt.subplots(figsize=(10,6))
sns.barplot(data=tips,x='day',y='total_bill')

ax.set_title('Average bill by day',fontsize=16)
ax.set_xlabel('Day of the week')
ax.set_ylabel('Average total bill')

plt.show()
```

```python
#Count plot
import seaborn as sns
import matplotlib.pyplot as plt
tips=sns.load_dataset('tips')
fig,ax=plt.subplots(figsize=(10,6))
sns.countplot(data=tips,x='day',hue='sex',ax=ax)
ax.set_title('Customer count by Day and Gender',fontsize=16)
ax.set_xlabel('Day of week')
ax.set_ylabel('Count')
plt.show()
```

```python
# Box plot
import seaborn as sns
import matplotlib.pyplot as plt
tips=sns.load_dataset('tips')
fig,ax=plt.subplots(figsize=(10,6))
sns.boxplot(data=tips,x='day',y='total_bill',ax=ax)
ax.set_title('Bill distrbution by day (Box plot)',fontsize=16)
ax.set_xlabel('Day of week')
ax.set_ylabel('Total bill')
plt.show()
```

```python
# Box plot
import seaborn as sns
import matplotlib.pyplot as plt
tips=sns.load_dataset('tips')
fig,ax=plt.subplots(figsize=(10,6))
sns.violinplot(data=tips,x='day',y='total_bill',ax=ax)
ax.set_title('Bill distrbution by day (Violin plot)')
ax.set_xlabel('Day of week')
ax.set_ylabel('Total bill')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips=sns.load_dataset('tips')
fig, (ax1,ax2)=plt.subplots(1,2,figsize=(14,6))
#box plot
sns.boxplot(data=tips,x='day',y='total_bill',ax=ax1)
ax1.set_title('Box Plot')
#violin plot
sns.violinplot(data=tips,x='day',y='total_bill',ax=ax2)
ax2.set_title('Violin Plot')
plt.tight_layout()
plt.show()
```
