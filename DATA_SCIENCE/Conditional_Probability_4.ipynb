```python
#2.Virus — Bayes Theorem
P_infected=0.002 #0.2%
P_positive_given_infected=0.98 #98%
P_positive_given_not_infected=0.01 #1%

P_not_infected=1-P_infected

P_positive=(
    P_positive_given_infected * P_infected + P_positive_given_not_infected * P_not_infected
)

P_infected_given_positive=(
    P_positive_given_infected * P_infected / P_positive

)

print("P(Infected | Positive)=",P_infected_given_positive)
print("percentage=",P_infected_given_positive * 100,'%')
```

```python
#3.Two Dice — Are A and B independent?
#Total outcomes when tosiing two dice
total=6*6

#event A: sum is even
A=[]

#event B:first die is 3
B=[]

#Find outcomes
for die1 in range(1,7):
    for die2 in range(1,7):

        if (die1+die2)%2==0:
            A.append((die1,die2))

        if die1==3:
            B.append((die1,die2))

# find A intersection B
A_and_B=set(A).intersection(set(B))

P_A=len(A)/total
P_B=len(B)/total
P_A_and_B=len(A_and_B)/total

print("P(A) =", P_A)
print("P(B) =", P_B)
print("P(A ∩ B) =", P_A_and_B)

# Check independence
if P_A_and_B == P_A * P_B:
    print("A and B are independent")
else:
    print("A and B are not independent")
```
