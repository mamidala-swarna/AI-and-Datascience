```python
#Set Theory and Logic
#-> order doesn't matter in sets
#-> sets doesn't accept duplicate elements

A={1,2,3}
C={1,1,2,3}
B={3,2,1}
```

```python
print(A,B,C)
```

```python
#important terms

#Universal Set ->All possible items
#Empty Set-> A set with no elements
#Cardinality of A ->Number of elements in A
```

```python
#Operations on sets
#Union of A,B -> elements in A or B -> for union we make use of pipe symbol |

A={1,2,3}
B={3,4,5}

print(A,B)
print(A|B) #using | pipe for Union
```

```python
#intersection of A,B ->elements that are in both A and B
print(A & B)
```

```python
#difference of A - B -> elements in A - (elements in common A and B)
#->elements in A but not in B

print(A-B)
print(B-A)
```

```python
#matplotlib -> helps us for visualization of data

import matplotlib.pyplot as plt
from matplotlib_venn import venn2

A=set([1,2,3,4])
B=set([3,4,5,6])

#plot venn diagram
venn2([A,B],set_labels=("A","B"))
plt.show()
```
