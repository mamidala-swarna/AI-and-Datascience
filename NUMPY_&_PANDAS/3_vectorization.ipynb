```python
#Vectorization and why Loops are slow
marks=[50,60,70,80,90]
new_marks=[]

for mark in marks:
    new_marks.append(mark+5)
print(new_marks)
```

```python
#Using numpy vectorization
import numpy as np
marks=np.array([50,60,70,80,90])
new_marks=marks+5
print(new_marks)
```

```python
#Multiplying Numbers
numbers=[1,2,3,4,5]
result=[]
for number in numbers:
    result.append(number*10)
print(result)
```

```python
#using numpy vectorization
numbers=np.array([1,2,3,4,5])
result=numbers*10
print(result)
```

```python
#Array Arithmetic
#Addition
a=np.array([10,20,30])
b=np.array([1,2,3])
result=a+b
print(result)
print("=" * 50)
#Substraction
a=np.array([100,200,300])
b=np.array([10,20,30])
result=a-b
print(result)
print("=" * 50)
#Multiplication
price=np.array([100,200,300])
print(price*4)
print("=" * 50)
#Division
numbers=np.array([10,20,30])
print(numbers/2)
print("=" * 50)
#Power
numbers=np.array([2,3,4,5])
print(numbers**2)
print("=" * 50)
#Modulus
numbers=np.array([10,11,12,13])
print(numbers%2)
```

```python
#Brodcasting Concepts
#Scalar broadcasting
numbers=np.array([10,20,30])
print(numbers+5)
```

```python
#giving 10% bonus
salary=np.array([20000,30000,40000])
new_salary=salary*1.10
print(new_salary)
```

```python
#Broadcasting with 2D arrays
matrix=np.array([
    [10,20,30],
    [40,50,60]
])
print(matrix+10)
```

```python
matrix=np.array([
    [10,20,30],
    [40,50,60]
])
bonus=np.array([1,2,3])
result=matrix+bonus
print(result)
```

```python
#Dot Product
a=np.array([1,2,3])
b=np.array([4,5,6])
result=np.dot(a,b)
print(result)
```

```python
#Shopping example
quantity=np.array([2,3,4])
price=np.array([10,20,30])
#calculate the total cost
total=np.dot(quantity,price)
print(total)
```

```python
#Matrix multiplication
A=np.array([
    [1,2],
    [3,4]
])
B=np.array([
    [5,6],
    [7,8]
])
result=np.matmul(A,B)
print(result)
```

```python
#@=>matrix multiplication
#*=>element eise multiplication
A = np.array([
    [1,2],
    [3,4]
])
B = np.array([
    [5,6],
    [7,8]
])

print("Element-Wise")
print(A*B)
print("=" * 50)
print("Matrix Multiplication")
print(A @ B)
```

```python
Marks=np.array([78,65,90,55,88,72,95,60])
print("Marks after adding Bonus:",Marks+5)
print("Average marks:",Marks.mean())
print("Highest marks:",Marks.max())
print("Lowest marks:",Marks.min())
print("Total marks:",Marks.sum())
print("marks greater than 75:",Marks[Marks>75])
print("Percentage:",(Marks/100)*100)
print("Sorted marks:",np.sort(Marks))
print("square of every mark:",Marks**2)
print(":","=" * 50)
Attendence=np.array([90,85,95,70,88,92,98,75])
print("Element-Wise")
print(Marks + Attendence)
```
