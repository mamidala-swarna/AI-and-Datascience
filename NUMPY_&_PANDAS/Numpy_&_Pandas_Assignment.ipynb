```python
#Reshape
arr=np.arange(1001,1013)
new_arr=arr.reshape(3,4)
print(new_arr)
```

```python
#first row
print(new_arr[0])
```

```python
#last column
print(new_arr[:,-1])
```

```python
#the first two rows
print(new_arr[0:2])
```

```python
#elements from the second row,columns 2 to 4
print(new_arr[1,1:4])
```

```python
#the last element using negative indexing
print(new_arr[-1,-1])
```

```python
#2x3 array of zeros
arr1=np.zeros((2,3))
print(arr1)
```

```python
#2x3 array of ones
arr2=np.ones((2,3))
print(arr2)
```

2.NumPy:Vectorization,Broadcasting & Dot Product

```python
quantity=np.array([2,3,4])
price=np.array([100,200,150])
```

```python
#calculating the total price for each product using vectorization
total_price=quantity+price
print(total_price)
```

```python
#adding 10% tax to every product using broadcasting
price=np.array([100,200,150])
final_price=price*1.10
print(final_price)
```

```python
#the final price after tax
print(final_price)
```

```python
#the total bill using the dot product
total_bill=np.dot(quantity,final_price)
print(total_bill)
```

```python
#the discount array
discount=np.array([10,20,15])
```

```python
#subtract the discount from the final prices
print(final_price-discount)
```

```python
#print the final amount for each product
A=np.array([[1,2],[3,4]])
B=np.array([[5,6],[7,8]])
```

```python
#perform element-wise multiplication using *
print("Element-Wise")
print(A * B)
```

```python
#perform matrix multiplication using @
print("Matrix Multiplication")
print(A @ B)
```

3.Pandas:Employee Data Analysis

```python
import pandas as pd
data = {
"Name": ["Anil", "Bhanu", "Charan", "Divya", "Esha"],
"Age": [22, 25, 21, 24, 23],
"City": ["Hyderabad", "Mumbai", "Hyderabad", "Bangalore", "Mumbai"],
"Salary": [30000, 45000, 25000, 50000, 40000]
}
df = pd.DataFrame(data)
```

```python
#Display the complete dataframe
print(df)
```

```python
#select only the name column
print(df["Name"])
```

```python
#Select name and salary
print(df[["Name","Salary"]])
```

```python
#select the row at index 2 using loc
print(df.loc[2])
```

```python
#Select rows 1 to 3 and columns Name and Salary using loc
print(df.loc[1:3,["Name","Salary"]])
```

```python
#elect rows at positions 1 to 3 and columns at positions 0 and 1 using iloc.
print(df.iloc[1:4,0:2])
```

```python
#find employees whose salary is greater than 30000
result=df[df["Salary"]>30000]
print(result)
```

```python
#Find employees whose Age is greater than 22 AND salary is greater than 35000
result=df[
    (df["Age"]>22) &
    (df["Salary"]>35000)
]
print(result)
```

```python
#Find the employess from Hyderabad OR Mumbai
result=df[
    (df["City"]== "Hyderabad") |
    (df["City"]=="Mumbai")
]
print(result)
```

```python
#Sort employees by salary in descending order
result=df.sort_values(by="Salary",ascending=False)
print(result)
```

```python
#Rename Salary as Monthly_Salary
result=df.rename(columns={"Salary":"Monthly_Salary"})
print(result)
```

```python
#Remove the City column
result=df.drop(columns="City")
print(result)
```

4.Pandas:Data Cleaning & Grouping

```python
import pandas as pd
data = {
"Name": ["Anil","Rahul","Priya","Sneha","Anil","Kiran",None,"Rahul"],
"Department": ["IT","HR","IT","Sales","IT","HR","Sales","HR"],
"Salary": [50000,40000,60000,None,50000,45000,55000,40000],
"Age": [25,28,24,30,25,None,26,28],
"City": ["Hyderabad","Delhi","Hyderabad","Mumbai","Hyderabad","Delhi","Mumbai","Delhi"],
"Joining_Date": ["2023-01-10","2022-05-15","2023-03-20","2021-07-12","2023-01-10","2022-08-25","2023-06-10","2022-05-15"]
}
df=pd.DataFrame(data)
print(df)
```

```python
#display df.info()
df.info()
```

```python
#display df.describe()
df.describe()
```

```python
#find the number of missing values in every column
print(df.isnull().sum().sum())
```

```python
#Identify missing values using isnull()
print(df.isnull())
```

```python
#Fill the missing Salary using the mean salary
mean_salary=df["Salary"].mean()
df["Salary"]=df["Salary"].fillna(mean_salary)
print(df["Salary"])
```

```python
#Fill the missing Age using the median age
median_age=df["Age"].median()
df["Age"]=df["Age"].fillna(median_age)
print(df["Age"])
```

```python
#If the City column contained a missing value, explain which strategy you would use for a categoricaln column
mode_city=df["City"].mode()[0]
df["City"]=df["City"].fillna(mode_city)
print(df["City"])
```

```python
#Find the number of employees in each department using value_counts()
print(df.value_counts("Department"))
```

```python
# Find the unique departments using unique().
print(df["Department"].unique())
```

```python
#Find the number of unique departments using nunique().
print(df["Department"].nunique())
```

```python
# Identify duplicate rows using duplicated()
print(df.duplicated())
```

```python
#Find the average salary for each department
print(df.groupby("Department")["Salary"].mean())
```

```python
#Find the total salary for each department.
print(df.groupby("Department")["Salary"].sum())
```

```python
#Find the maximum and minimum salary for each department.
result=df.groupby("Department")["Salary"].agg(
    ["max","min"]
)
print(result)
      
```

```python
#Calculate sum, mean, max, and min together using agg().
result=df.groupby("Department")["Salary"].agg(
    ["sum","mean","max","min"]
)
print(result)
```
