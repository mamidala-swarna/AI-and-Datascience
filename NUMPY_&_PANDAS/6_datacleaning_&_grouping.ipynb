```python
import pandas as pd

data = {
"Name": ["Anil", "Rahul", "Priya", "Sneha", "Anil", "Kiran", None,"Rahul"],
"Department": ["IT", "HR", "IT", "Sales", "IT", "HR", "Sales", "HR"],
"Salary": [50000, 40000, 60000, None, 50000, 45000, 55000, 40000],
"Age": [25, 28, 24, 30, 25, None, 26, 28],
"City": ["Hyderabad", "Delhi", "Hyderabad", "Mumbai", "Hyderabad",
"Delhi", "Mumbai", "Delhi"],
"Joining_Date": [
"2023-01-10",
"2022-05-15",
"2023-03-20",
"2021-07-12",
"2023-01-10",
"2022-08-25",
"2023-06-10",
"2022-05-15"
]
}
df = pd.DataFrame(data)
print(df)
```

```python
#Grouping data with groupby()
#groupby()=>is used to divide data into groups based on one or more columns

#groupby one column
groups = df.groupby("Department")
print(groups)
```

```python
print(df.groupby("Department")["Salary"].mean())
```

```python
#groupby multiple columns
result=df.groupby(["Department","City"])["Salary"].mean()
print(result)
```

```python
#groupby()+aggregation
result=df.groupby("Department")["Salary"].sum()
print(result)
```

```python
#multiple aggregates
result=df.groupby("Department")["Salary"].agg(
    ["sum","mean","max","min"]
)
print(result)

#groupby() decides which groups to create
#aggregation decides what calculation to perform on each group
```

```python
#Aggregation Functions
#Aggregation=>summarizing multiple values into a single value
```

```python
#Handling Missing Values
#isnull()
print(df["Salary"].isnull())
```

```python
#counting missing values
print(df.isnull().sum().sum())
```

```python
#dropna()=>used to remove rows containing missing values
new_df=df.dropna()
print(new_df)
```

```python
#Original dataframe is not changed unless you use
#new_df=df.dropna(inplace=True)
#print(new_df)
```

```python
#fillna()=>used to replace missing values
df["Salary"]=df["Salary"].fillna(50000)
print(df["Salary"])
```

```python
#mean strategy
mean_salary=df["Salary"].mean()
df["Salary"]=df["Salary"].fillna(mean_salary)
```

```python
#median strategy
median_age=df["Age"].median()
print("Median Age:",median_age)
df["Age"]=df["Age"].fillna(median_age)
```

```python
#mode strategy
mode_city=df["City"].mode()[0]
df["City"]=df["City"].fillna(mode_city)
```

```python
#value_counts=>counts how many times each value occurs
print(df["Department"].value_counts())
```

```python
#unique()
print(df["Department"].unique())
```

```python
#nunique()=>returns the number of unique values
print(df["Department"].nunique())
```

```python
#removing duplicates=>A duplicate is a row that apperas more than once
#duplicated()=>identifies duplicate rows
print(df.duplicated())
```

```python
#drop_duplicates()=>remove duplicate rows
new_df=df.drop_duplicates()
print(new_df)
```

```python
#datatype conversion
#astype()=>used to convert data type
df["Age"]=df["Age"].astype("float")
```

```python
#to_numeric()=>used to convert values into numeric data
df["Salary"]=pd.to_numeric(df["Salary"],errors="coerce")
```

```python
data=["50000","45000","abc"]
result=pd.to_numeric(data,errors="coerce")
print(result)
```

```python
#to_datetime()=>
#used toconvert strings into actual datetime values
print(df["Joining_Date"].dtype)
```

```python
#convert
df["Joining_Date"]=pd.to_datetime(df["Joining_Date"])
```

```python
print(df["Joining_Date"].dtype)
```

```python
#extract year
df["Year"]=df["Joining_Date"].dt.year
print(df[["Joining_Date","Year"]])
```

```python
#extract month
df["Month"]=df["Joining_Date"].dt.month
print(df[["Joining_Date","Month"]])
```

```python
#extract day
#Data filtering=>once converted to datetime,we can filter based on dates
#finding the employees who joined after 1 january,2023
result=df[df["Joining_Date"]>"2023-01-01"]
print(result)
```

```python
#filter by year
result=df[df["Joining_Date"].dt.year==2023]
print(result)
```
