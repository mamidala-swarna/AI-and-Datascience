```python
import pandas as pd
data={
    "Name":["Swarna","Vighnesh","Keerthana","Sneha","Shruthi"],
    "Age":[21,22,20,24,23],
    "City":["Hyderabad","Mumbai","Hyderabad","Bangalore","Mumbai"],
    "Salary":[40000,50000,30000,45000,50000]
}
df=pd.DataFrame(data)
print(df)
```

```python
print(df["Name"])
```

```python
print(df[["Name","Salary"]])#when we selecting multiple columns,we use double square brackets
```

```python
#loc[] vs iloc[]=>are used to select rows and columns
#loc[]=>selects using labels/names
#iloc=>selects using integer positions
#loc[]
print(df.loc[2])
```

```python
print(df.loc[1:3,["Name","Salary"]])#in loc[] the ending index is also been included
```

```python
#iloc[]
print(df.iloc[2])
```

```python
print(df.iloc[1:4,0:2])#it doesnot include ending index
```

```python
df.loc[1:3]
```

```python
df.iloc[1:3]
```

```python
#Boolean Filtering=>is used to select rows based on a condition
result=df[df["Salary"]>30000]
print(result)
```

```python
result=df[
    (df["Age"]>22) &
    (df["Salary"]>35000)
]
print(result)
```

```python
result=df[
    (df["City"]=="Hyderabad") |
    (df["City"]=="Mumbai")
]
print(result)
```

```python
#isin()=>used to check whether values belong to a specified list of values
result=df[df["City"].isin(["Hyderabad","Mumbai"])]
print(result)
```

```python
#between()=>checks whether values fall within a specified range
#by default,both boundary values are included
result=df[df["Age"].between(22,24)]
print(result)
```

```python
#Query()=>allows us to filter a dataframe using a string expression
#it cam make conditions more readable
result=df.query("Salary>30000")
print(result)
```

```python
result=df.query("Age>22 and Salary>35000")
print(result)
```

```python
#sort_values()=>sort a dataframe based on the values in one or more columns
result=df.sort_values(by="Salary") #ascending order
print(result)
```

```python
#sorting in descending order
result=df.sort_values(
    by="Salary",
    ascending=False
)
print(result)
```

```python
#sort_index()=>used to sort datframe based on its index value
new_df=df.iloc[[3,1,4,0,2]]
print(new_df)
```

```python
#sort by index
result=new_df.sort_index()
print(result)
```

```python
#adding column and rows
df["Experience"]=[1,2,0,3,3]
print(df)
```

```python
#adding a calculated column
df["Bonus"]=df["Salary"]*0.10
print(df)
```

```python
#adding a new row
df.loc[len(df)]=["Hani",21,"Pune",45000,3,4500]#len(df)=>returns the next avilable row position
print(df)
```

```python
result=df.rename(
    columns={
        "Name":"Employee_Name",
        "Salary":"Monthly_Salary"
    }
)
print(result)
```

```python
#renaming rows
result=df.rename(
    index={
        0:"A",
        1:"B",
        2:"C"
    }
)
print(result)
```

```python
#dropping columns and rows
#drop()=>is used to remove rows or columns from a dataframe
result=df.drop(columns="City")
print(result)
```

```python
#multiple columns
result = df.drop(columns=["City", "Age"])
print(result)
```

```python
#dropping a row
result=df.drop(index=5)
print(result)
```

```python
#dropping multiple rows
result=df.drop(index=[3,4])
print(result)
```
