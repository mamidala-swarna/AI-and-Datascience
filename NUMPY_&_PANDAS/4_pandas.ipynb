Pandas

```python
import pandas as pd

#Series=>A pandas series is one-dimensional labeled data structure

numbers=[10,20,30,40,50]
series=pd.Series(numbers)
series
```

```python
#Acessing series values
print(series[0])
print(series[1])
```

```python
#creating a series with custom index
marks=[85,90,78]
student_marks=pd.Series(
    marks,
    index=["Ananya","Swarna","Harini"]
)
student_marks
```

```python
student_marks["Swarna"]
```

```python
#series from a dictionary
#keys become indexes and values become series values
student={
    "name":"Swarna",
    "age":22,
    "role":"AI-Ml Engineer"
}
data=pd.Series(student)
data
```

```python
#DatFrame=> A pandas dataframe is a two-dimensional tabular data structure consisting of rows and columns
#Creating dataFrames
data={
    "Name":["Ananya","Swarna","Harini"],
    "Age":[22,21,22],
    "Course":["Java","Data Science","Full Stack"]
}
df=pd.DataFrame(data)
df
```

```python
#creating a df using a list of lists
data=[
    ["Ananya",22,"Java"],
    ["Swarna",21,"Data Science"],
    ["Harini",22,"Full Stack"]
]
df=pd.DataFrame(
    data,
    columns=["Name","Age","Course"]
)
df
```

```python
#creating a df from a list of dictionaries
data=[
    {
        "Name":"Ananya",
        "Age":22,
        "Course":"Java"
    },
    {
        "Name":"Swarna",
        "Age":21,
        "Course":"Data Science"

    },
    {
        "Name":"Harini",
        "Age":22,
        "Course":"Full Stack"
    }
]
df=pd.DataFrame(data)
df
```

```python
#creating an empty dataframe
df=pd.DataFrame()
df
```

```python
#Reading data using read_csv()
#pd.read_csv("filename.csv")=>reads the CSV filr and converts it into a pandas dataframe
df=pd.read_csv("employees.csv")
df
```
